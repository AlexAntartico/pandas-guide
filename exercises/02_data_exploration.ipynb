### 6. A reusable quick-EDA function

```python
def quick_eda(df):
    print(f'Shape:       {df.shape}')
    print(f'Missing:     {df.isnull().sum().sum()}')
    print(f'Duplicates:  {df.duplicated().sum()}')
    return df.describe(include='all')

quick_eda(orders)
```
