```python
import pandas as pd

orders = pd.read_csv('datasets/raw/orders.csv', parse_dates=['order_date'])
customers = pd.read_csv('datasets/raw/customers.csv')
```

### 1. Indexing — loc vs iloc

```python
orders.iloc[0]            # first row by position
orders.loc[0, 'order_id']  # single value by label
```

### 2. Boolean filtering

```python
completed = orders[orders['status'] == 'completed']
completed.shape
```

```python
# query() for readability
orders.query("status == 'completed' and total > 1000").head()
```

```python
# isin() for multiple values
orders[orders['payment_method'].isin(['credit_card', 'paypal'])].shape
```

### 3. Sorting

```python
orders.sort_values('total', ascending=False).head(10)
```

### 4. GroupBy — split-apply-combine

```python
orders.groupby('status')['total'].agg(['count', 'sum', 'mean'])
```

```python
customers.groupby('membership')['customer_id'].count()
```

### 5. Merge across tables

```python
order_items = pd.read_csv('datasets/raw/order_items.csv')
products = pd.read_csv('datasets/raw/products.csv')

# orders -> customers -> order_items -> products
merged = (orders
    .merge(customers, on='customer_id')
    .merge(order_items, on='order_id')
    .merge(products, on='product_id'))
merged.shape
```

```python
# Revenue by product category
merged.groupby('category')['line_total'].sum().sort_values(ascending=False)
```

### 6. Concat (stack rows)

```python
pd.concat([orders.head(3), orders.tail(3)])
```
