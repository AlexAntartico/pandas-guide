```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('deep')

orders = pd.read_csv('datasets/raw/orders.csv', parse_dates=['order_date'])
customers = pd.read_csv('datasets/raw/customers.csv')
order_items = pd.read_csv('datasets/raw/order_items.csv')
products = pd.read_csv('datasets/raw/products.csv')
```

### 1. Monthly revenue — line chart

```python
monthly = orders.set_index('order_date')['total'].resample('ME').sum()
monthly.plot(figsize=(12, 4), title='Monthly Revenue')
```

### 2. Revenue by category — bar chart

```python
by_cat = (orders.merge(order_items, on='order_id')
             .merge(products, on='product_id')
             .groupby('category')['line_total'].sum()
             .sort_values())
by_cat.plot(kind='barh', figsize=(10, 6), title='Revenue by Category')
```

### 3. Order totals — histogram

```python
orders['total'].plot(kind='hist', bins=50, figsize=(10, 5),
                       title='Order Total Distribution')
```

### 4. Scatter — quantity vs line total

```python
order_items.plot(kind='scatter', x='quantity', y='line_total',
                  alpha=0.3, figsize=(8, 6))
```

### 5. Box plot — total by membership

```python
sns.boxplot(data=orders.merge(customers, on='customer_id'),
            x='membership', y='total')
```

### 6. Correlation heatmap

```python
sns.heatmap(orders.select_dtypes(include='number').corr(),
            annot=True, cmap='coolwarm', center=0)
```
